# Cleaning our data

Our dataset is downloaded from a website that tracks a lot of sports betting outcomes, and this information is not something I want in my dataset.

In order to clean the data for this project I will be picking the columns I wish to work with and moving them into `cleaned_premier.csv`

First, we must import `premier_data_2019-2026.csv`

```python
import pandas as pd

df_premier = pd.read_csv("../data/premier_data_2019-2026.csv")

df_premier.head()
```

Next step is to pick the columns we wish to save.

We do this by putting them into a `keep_columns` list

```python
keep_columns = [
    "Div",
    "Date",
    "Time",
    "HomeTeam",
    "AwayTeam",
    "FTHG",
    "FTAG",
    "FTR",
    "HTHG",
    "HTAG",
    "HTR",
    "Referee",
    "HS",
    "AS",
    "HST",
    "AST",
    "HF",
    "AF",
    "HC",
    "AC",
    "HY",
    "AY",
    "HR",
    "AR"
]
```

Lastly, we apply our `keep_columns` list to our data frame and import it to our data folder

```python
cleaned_premier = df_premier[keep_columns]

cleaned_premier.to_csv(
    "../data/cleaned_premier.csv",
    index=False
)
```
