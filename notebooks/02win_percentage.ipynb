# Win and Point Percentage

Separating our games based on if a team is playing at home or away and finding win percentage.
Adding a point percentage, which includes all times a team got a win or a draw.

```python
standings = standings.sort_values(
    "Home Win Percentage",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(standings))
width = 0.35

fig, ax = plt.subplots(figsize=(12,6))

ax.bar(x - width/2, standings["Home Win Percentage"], width, label = "Home", color = "#2a459a")
ax.bar(x + width/2, standings["Road Win Percentage"], width, label = "Road", color = "#d73c23")

ax.set_xticks(x)
ax.set_xticklabels(standings["Team"], rotation=45, ha="right")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.set_ylabel("Win Percentage")
ax.set_title("Home vs Road Win Percentage")
ax.legend()

plt.tight_layout
plt.show()
```
