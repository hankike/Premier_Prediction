```python
url = "https://raw.githubusercontent.com/hankike/Premier_Prediction/refs/heads/main/data/cleaned_premier.csv"

corners = create_corners_table(url)
```

# Corner Kicks

Corner kicks can set an offense in action

```python
corners
```

```python
corners = corners.sort_values(
    "Corners Home per Game",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(corners))
width = 0.35

fig, ax = plt.subplots(figsize=(12,6))

ax.bar(x - width/2, corners["Corners Home per Game"], width, label = "Home", color = "#6314b5")
ax.bar(x + width/2, corners["Corners Allowed Home per Game"], width, label = "Road", color = "#f25050")

ax.set_xticks(x)
ax.set_xticklabels(corners["Team"], rotation=45, ha="right")
ax.set_ylabel("Corners per Game")
ax.set_title("Corners Had vs Allowed at Home")
ax.legend()

plt.tight_layout
plt.show()
```

```python
corners = corners.sort_values(
    "Corners Away per Game",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(corners))
width = 0.35

fig, ax = plt.subplots(figsize=(12,6))

ax.bar(x - width/2, corners["Corners Away per Game"], width, label = "Home", color = "#6314b5")
ax.bar(x + width/2, corners["Corners Allowed Away per Game"], width, label = "Road", color = "#f25050")

ax.set_xticks(x)
ax.set_xticklabels(corners["Team"], rotation=45, ha="right")
ax.set_ylabel("Corners per Game")
ax.set_title("Corners Had vs Allowed on the Road")
ax.legend()

plt.tight_layout
plt.show()
```
