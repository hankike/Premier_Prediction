```python
url = "https://raw.githubusercontent.com/hankike/Premier_Prediction/refs/heads/main/data/cleaned_premier.csv"

fouls = create_fouls_table(url)
```

# Fouls

```python
fouls
```

```python
average = fouls['Booking Points per Game'].mean()

fouls['Difference'] = fouls['Booking Points per Game'] - average

fouls_sorted = fouls.sort_values('Difference')

fig, ax = plt.subplots(figsize = (10, 8))

ax.barh(fouls_sorted['Team'], fouls_sorted['Difference'])

ax.axvline(0, color = 'black', linewidth = 1)

ax.set_xlabel('Difference from League Average')
ax.set_title('Booking Points Difference from Average')
```

```python
average = fouls['Yellows per Game'].mean()

fouls['Difference'] = fouls['Yellows per Game'] - average

fouls_sorted = fouls.sort_values('Difference')

fig, ax = plt.subplots(figsize = (10, 8))

ax.barh(fouls_sorted['Team'], fouls_sorted['Difference'])

ax.axvline(0, color = 'black', linewidth = 1)

ax.set_xlabel('Difference from League Average')
ax.set_title('Yellows Difference from Average')
```
