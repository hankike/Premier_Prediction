```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path
```

```python
sys.path.append(str(Path.cwd().parent))

from src.tables import create_shots_table

from src.tables import create_shots_against_table
```

```python
url = "https://raw.githubusercontent.com/hankike/Premier_Prediction/refs/heads/main/data/cleaned_premier.csv"

shots = create_shots_table(url)

shots_against = create_shots_against_table(url)
```

# Comparing Our Data

Now we want to look at a visual showing how the teams relate to shots per goal and shots per game, both for and against them

```python
shots = shots.sort_values(
    "Home Shots per Game",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(shots["Games"]))

width = 0.35

plt.figure(figsize=(14, 7))

# Home bar
plt.bar(
    x - width/2,
    shots["Home Shots per Game"],
    width=width,
    color = '#648fff',
    label="Home Shots per Game"
)

plt.bar(
    x - width/2,
    shots["Home Shots per Goal"],
    width=width,
    color = '#274bfc',
    label="Home Shots per Goal"
)

# Away bar
plt.bar(
    x + width/2,
    shots["Away Shots per Game"],
    width=width,
    color = '#ff007d',
    label="Away Shots per Game"
)

plt.bar(
    x + width/2,
    shots["Away Shots per Goal"],
    width=width,
    color = '#98165d',
    label="Away Shots per Goal"
)

plt.xlabel("Team")
plt.ylabel("Shots")
plt.title("Premier League Shots 2019 - 2026")

plt.xticks(x, shots["Team"], rotation=45, ha="right")
plt.yticks(np.arange(0, 21, 2))

plt.legend()

plt.tight_layout()
plt.show()
```

```python
shots_against = shots_against.sort_values(
    "Home Opp Shots per Game",
    ascending=True
).reset_index(drop=True)

x = np.arange(len(shots_against["Games"]))

width = 0.35

plt.figure(figsize=(14, 7))

# Home bar
plt.bar(
    x - width/2,
    shots_against["Home Opp Shots per Game"],
    width=width,
    color = '#67e18d',
    label="Home Opp Shots per Game"
)

plt.bar(
    x - width/2,
    shots_against["Home Opp Shots per Goal"],
    width=width,
    color = '#1bb999',
    label="Home Opp Shots per Goal"
)

# Away bar
plt.bar(
    x + width/2,
    shots_against["Away Opp Shots per Game"],
    width=width,
    color = '#f79e70',
    label="Away Opp Shots per Game"
)

plt.bar(
    x + width/2,
    shots_against["Away Opp Shots per Goal"],
    width=width,
    color = '#fe6100',
    label="Away Opp Shots per Goal"
)

plt.xlabel("Team")
plt.ylabel("Shots")
plt.title("Premier League Shots Against 2019 - 2026")

plt.xticks(x, shots_against["Team"], rotation=45, ha="right")
plt.yticks(np.arange(0, 21, 2))

plt.legend()

plt.tight_layout()
plt.show()
```
