```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path
```

```python
sys.path.append(str(Path.cwd().parent))

from src.tables import create_results_table
```

```python
url = "https://raw.githubusercontent.com/hankike/Premier_Prediction/refs/heads/main/data/cleaned_premier.csv"

results = create_results_table(url)
```

# Team Records 

```python
results = results.sort_values(
    "Wins",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(results["Games"]))
width = 0.25

plt.figure(figsize=(14, 7))

plt.bar(
    results["Team"],
    results["Wins"],
    label="Wins"
)

plt.bar(
    results["Team"],
    results["Losses"],
    bottom=results["Wins"],
    label="Losses"
)

plt.bar(
    results["Team"],
    results["Draws"],
    bottom=results["Wins"] + results["Losses"],
    label="Draws"
)
plt.xlabel("Team")
plt.ylabel("Games Played")
plt.title("Premier League Record Splits, 2019 - 2026")
plt.xticks(x, results["Team"], rotation=45, ha="right")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
results = results.sort_values(
    "Wins",
    ascending=False
).reset_index(drop=True)

x = np.arange(len(results["Team"]))
width = 0.25

plt.figure(figsize=(14, 7))

plt.bar(x - width, results["Wins"], width, label="Wins")
plt.bar(x, results["Losses"], width, label="Losses")
plt.bar(x + width, results["Draws"], width, label="Draws")

plt.xlabel("Team")
plt.ylabel("Number of Games")
plt.title("Premier League Wins/Losses/Ties, 2019 - 2022")
plt.xticks(x, results["Team"], rotation=45, ha="right")
plt.legend()

plt.tight_layout()
plt.show()
```
